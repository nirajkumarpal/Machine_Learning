# Relational Plot

## plots under this section

* lineplot

```python
tips=sns.load_dataset('tips')
tips
```

```python
# scatter plot -> axes level function
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip',hue='sex',style='time',size='size')
```

```python
# relplot -> figure level function -> square shape
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size = 'size')
```

```python
# scatter using relplot -> size and hue
```

```python
# style semantics
```

```python
# line plot
gap=px.data.gapminder()
temp_df=gap[gap['country']=='India']
temp_df
```

```python
# axes level function
sns.lineplot(data=temp_df,x='year',y='lifeExp',)
```

```python
# relplot 
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

```python
# using relplot 
```

```python
# # hue -> style
temp_df=gap[gap['country'].isin(['India','Brazil','Pakistan','South Africa'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
# facet plot -> figure level function ->work with relplt
# it will not work with axes level fuction
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',row='day',col='sex')
```

```python
# col wrap
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=3)
```

```python
# plotting univariate histogram
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
# bins parameter
sns.displot(data=tips,x='total_bill',kind='hist',bins=20)
```

```python
# countplot 
sns.displot(data=tips,x='day',kind='hist',)
```

```python
# elements -> step
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',hue='sex',element='step')
```

```python
# faceting using col and row -> not work on histogram functions
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step',col='day',col_wrap=3)
```

```python
# kde plot -> kernal density estimation
# rather than using discrete bins , a KDE plot smooths the observations with a Gaussian kernel to create a continuous density estimate.
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill =True)
```

```python
# Rugplot
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histogram
sns.histplot(data=tips,x='total_bill',y='tip')
sns.displot(data=tips,x='total_bill',y='tip',kind='hist',bins=20)
```

```python
# Bivariate KDE plot
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

# Matrix plot
* Heatmap
* clustermap

```python
# headmap

# plot rectangular data as a color-encoded matrix
temp_df=gap.pivot(index='country',columns='year',values='lifeExp')
temp_df
```

```python
# axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
# annot
temp_df=gap[gap['continent'] =='Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidths=0.5,cmap='YlGnBu')
```
