## Annotations

```python
batters = pd.read_csv('/Users/nirajkumarpal/Desktop/DATA_SCIENCE/session-datasets/datasets-session-24/batter.csv')
batters.shape
```

```python
sample_df = batters.head(100).sample(25,random_state=5)
```

```python
sample_df
```

```python
plt.figure(figsize = (18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
x = [1,2,4,5,6]
y = [7,8,9,10,11]
plt.scatter(x,y)
plt.text(1,7,'point 1')
plt.text(2,8,'point 2')
plt.text(4,9,'point 3')
plt.text(5,10,'point 4',fontdict = {'size':12,'color':'red'})
```

```python
# horizontal and vertical line
plt.figure(figsize = (18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

plt.axhline(130,color = 'red')
plt.axhline(140,color = 'green')
plt.axvline(30,color = 'red')
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```
