# parametric density Estimation

```python
import matplotlib.pyplot as plt
import numpy as np
from numpy.random import normal
```

```python
sample = normal(loc=50,scale=5,size=100)
```

```python
sample.mean()
```

```python
# plot histogram to understand the distributation of data
plt.hist(sample,bins=10)
```

```python
# calculate sample mean and sample std dev
sample_mean = sample.mean()
sample_std = sample.std()
```

```python
# fit the distributaion with the above parameter
from scipy.stats import norm
dist = norm(loc=sample_mean, scale=sample_std)
```

```python
values = np.linspace(sample.min(),sample.max(),100)
```

```python
sample.max()
```

```python
probabilities = [dist.pdf(value) for value in values]
```

```python
# plot the histogram and pdf
plt.hist(sample,bins=10#Eeefjdhddddhdhfhfhfhffjdjdjddddddhdhdhdhhdhdhdhdddd#33#dddfffjdjdjdjdffjfjfjffffffffdaddddddd,density = True)
plt.plot(values,probabilities)
```

# KDE

```python
# generate a sample
sample1 = normal(loc=20, scale=5, size=300)
sample2 = normal(loc=40, scale=5, size=700)
sample = np.hstack((sample1, sample2))
```

```python
sample
```

```python
# plot histogram bins=50
plt.hist(sample,bins=50)
```

```python
from sklearn.neighbors import KernelDensity
```

```python
# convert data to a 2D array
sample = sample.reshape(len(sample), 1)
model = KernelDensity(kernel='gaussian', bandwidth=3.0)
model.fit(sample)
```

```python
values = np.linspace(sample.min(),sample.max(),100)
values = values.reshape(len(values),1)
```

```python
probabilities = model.score_samples(values)
probabilities = np.exp(probabilities)
```

```python
plt.hist(sample,bins=50,density = True)
plt.plot(values[:],probabilities)
plt.show()
```

```python
import seaborn as sns
kde_sample = np.hstack((sample1, sample2))
sns.kdeplot(kde_sample, bw_adjust=0.2)

```
